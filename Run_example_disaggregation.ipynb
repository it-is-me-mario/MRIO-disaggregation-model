```python
import mario
```

Read the aggregated version of GTAP12 (2 regions, 4 sectors, 1  consumption category, 1 value added category)

```python
db_path= #your path to GTAP12_example_db.xlsx
db=mario.parse_from_excel(db_path,
                          table='IOT',
                          mode='flows')
```

Read the inventory file and disaggregate the tables

```python
inventory_path= #your path to Inventory_toy.xlsx
db.read_add_sectors_excel(inventory_path,
                          split=True,
                          read_inventories=True)

cvxlab_path = #where you wish to have the disaggregation auxiliary folder located 
split_db = db.add_sectors(inplace=False,split=True,
                    cvxlab_path=cvxlab_path,
                    parent_names={'Cement': 'Secondary_rest'}, 
                    solver='CLARABEL', #better MOSEK, free academic licence available
                    residue=1e-6) #Cells under 1 USD are considered 0
```

The enhanced GTAP database is contained in split, and can be explored

Export the whole table to an excel file

```python
desired_output_path= #where you wish to have the excel output file located
split.to_excel(desired_output_path)
```
